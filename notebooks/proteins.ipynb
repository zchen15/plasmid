```python
import molviewspec

filename = "../data/pdb/4ZT9.cif"

# generate the view
builder = molviewspec.create_builder()
structure = builder.download(url=filename).parse(format='mmcif').model_structure()

# custom drawing
structure.component(selector='polymer').representation(type='cartoon').color(color="green")

structure.component(selector='ligand').representation(type='ball_and_stick').color(
    custom={"molstar_color_theme_name": "element-symbol"}
)

# render to the notebook
state = builder.get_state()
molviewspec.molstar_notebook(state)
```
