## Loading a pdb file
The PDB module can be used to read and alter PDB structure files.

```python
rec = read_pdb("../data/pdb/4ZT9.cif")
rec.reset_index()

c = rec.atom_array["chain_id"] == "A"
c|= rec.atom_array["chain_id"] == "B"
rec = rec[c]

c = rec.atom_array["chain_id"] == "A"
rec.annotate_by_atom_id(atom_id = rec.atom_array.loc[c, "atom_id"],
                        name = "dCas9",
                        color = "orange",
                        inplace = True,
                       )
c = rec.atom_array["chain_id"] == "B"
rec.annotate_by_atom_id(atom_id = rec.atom_array.loc[c, "atom_id"],
                        name = "gRNA",
                        color = "blue",
                        inplace = True,
                       )
rec
```

## Viewing with Molstar

```python
molv = Viewer(rec)
view = molv.draw()
view
```
